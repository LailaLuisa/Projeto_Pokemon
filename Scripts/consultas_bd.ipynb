Configuração básica do ambiente para realização das consultas

```python
import os
import psycopg2
from dotenv import load_dotenv

load_dotenv()

dbname = os.getenv("db_name")
user = os.getenv("db_user")
password = os.getenv("db_password")
host = os.getenv("db_host")
port = os.getenv("db_port")

# Criar conexão

conexao = psycopg2.connect(dbname = dbname, user = user, password = password, host = host, port = port)

cursor = conexao.cursor() 
```

Algumas consultas ao banco com comandos SQL através do cursor

```python
cursor.execute('SELECT * FROM pokemon_info WHERE id = 130')
linhas = cursor.fetchall()

for linha in linhas:
  print(linha)
```

```python
cursor.execute(
    """
    SELECT * FROM pokemon_info WHERE tipo_principal = 'fire'
    """
)

linhas = cursor.fetchall()

for linha in linhas:
  print(linha)
```

```python
cursor.execute(
    """
    SELECT * FROM pokemon_info WHERE nome LIKE 'b%'
    """
)

linhas = cursor.fetchall()

for linha in linhas:
  print(linha)
```
